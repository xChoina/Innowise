```python
import numpy as np
```

**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

```python
arr = np.random.randint(10,size=(10))
print(arr)
arr[(arr>3) & (arr<8)] *= -1
print(arr)
```

**2.** Replace the maximum element of a random array with 0

```python
arr = np.random.randint(10,size=(10))
print(arr)
arr[arr==max(arr)] = 0
print(arr)
```

**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
arr = np.random.randint(10,size=(3,2))
print(arr)
print('')
arrp = np.meshgrid(*arr, indexing='ij')
res = np.stack(arrp,axis=-1).reshape(-1,arr.shape[0])
print(res)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
A = np.random.randint(10,size=(8,3))
B = np.random.randint(10,size=(2,2))
print(A)
print(B)
print("-" * 30)
for i, pom1 in enumerate(A):
  for j, pom2 in enumerate(B):
    if set(pom2).issubset(set(pom1)):
      print(f" Row {pom1} have elements of row {pom2}")
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
arr=np.random.randint(5, size=(10,3))
print(arr)
print("-" * 30)
res=[]
for i in range(len(arr)):
    if arr[i].min() != arr[i].max():
      res.append(arr[i])

print(np.array(res))
```

**6.** Given a two-dimensional array, remove the rows that are repeated.






```python
arr = np.random.randint(10,size=(10,3))
print(arr)
print('-'*30)
res = np.unique(arr,axis=0)
print(res)
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
#without numpy
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
print(X)
shorter = min(len(X), len(X[0]))
prod = 1
for i in range(shorter):
  if X[i][i] != 0:
    prod*=X[i][i]

print("Product of nonzero elements: ",prod)
```

```python
#with numpy
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
res = np.diag(X)
print(np.prod(res[res !=0]))
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero.
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
#withoutnp
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
maxi = 0
for i in range(len(x)-1):
  if x[i] == 0:
    maxi = max(maxi,x[i+1])

print(maxi)
```

```python
#withnp
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
res = np.max(x[1:][x[:-1]== 0])
print(res)
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
#withoutnp
x = np.array([2, 2, 2, 3, 3, 3, 5])
unix = []
dupx = []
cc=1
curr=x[0]
for i in range(1,len(x)):
  if x[i] == curr:
    cc+=1
  else:
    unix.append(curr)
    dupx.append(cc)
    curr = x[i]
    cc=1
unix.append(curr)
dupx.append(cc)
print(unix)
print(dupx)
```

```python
#withnp
x = np.array([2, 2, 2, 3, 3, 3, 5])
res = np.unique_counts(x)
print(res)
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
#withoutnp
import math
X=np.random.randint(20,size=(2,2))
Y=np.random.randint(20,size=(3,2))
res = []
print(X)
print(Y)
print("-"*30)
for i in X:
  row = []
  for j in Y:
    sqr_diff = sum((a-b) ** 2 for a, b in zip(i,j))
    distance = math.sqrt(sqr_diff)
    row.append(distance)
  res.append(row)
for i in res:
  print(i)
```

```python
#withnp
import time
from scipy.spatial.distance import cdist
X=np.random.rand(1000,2)
Y=np.random.rand(1000,2)
print(f"X: {X}")
print(f"Y: {Y}")
start = time.time()
distance = np.sqrt(np.sum((X[:,np.newaxis,:]-Y) ** 2,axis=2))
ntime = time.time() - start
print(f"Resuslt np: {distance}")
start = time.time()
distance_cdist = cdist(X,Y)
ctime = time.time() - start
print(f"Result scipy: {distance_cdist}")

print(f"Czas numpy: {ntime:.5f}")
print(f"Czas scipy: {ctime:.5f}")
```

You work in the marketing department of a food company called MyCrunch, which is developing a new type of delicious, healthy cereal called **CrunchieMunchies**.

You want to show consumers how healthy your cereal is compared to other leading brands, so you've compiled nutrition data from several different competitors.

Your task is to use Numpy calculations to analyze this data and prove that your **СrunchieMunchies** - the healthiest choice for consumers.


```python
import numpy as np
```

1. View the cereal.csv file. This file contains calorie counts for various brands of cereal. Download the data from the file and save it as calorie_stats.

```python
calorie_stats = np.loadtxt("/cereal.csv", delimiter=",")
calorie_stats
```

2. One serving of CrunchieMunchies contains 60 calories. How much higher is the average calorie count of your competitors?

Store the answer in the variable average_calories and print the variable in the terminal

```python
average_calories = np.average(calorie_stats)
print(f"Average calories of other companies is {average_calories}")
print(f"This is {round(average_calories-60,2)} higher than CrunchieMunchies")
```

3. Does the average calorie count accurately reflect the distribution of the data set? Let's sort the data and see.

Sort the data and store the result in the calorie_stats_sorted variable. Print the sorted information.

```python
sorted_calories = np.sort(calorie_stats)
print(f"Sorted calories: {sorted_calories}")
```

4. It looks like most values ​​are above the mean. Let's see if the median is the most accurate measure of the data set.

Calculate the median of the data set and save your answer in median_calories. Print the median so you can see how it compares to the mean.

```python
median_calories = np.median(calorie_stats)
print(f"Median value: {median_calories}")
```

5. While the median shows that at least half of our values ​​are over 100 calories, it would be more impressive to show that a significant portion of the competitors have higher calorie counts than CrunchieMunchies.

Calculate various percentiles and print them until you find the lowest percentile that exceeds 60 calories. Store this value in the nth_percentile variable.

```python
for i in range(1,100):
  actual_p = np.percentile(calorie_stats,[i])
  print(f"Percentil {i} totals: {actual_p}")
  if actual_p > 60:
    nth_percentile = i
    print(f"The lowest percentile is: {nth_percentile}")
    break
```

6. While the percentile shows us that most competitors have much higher calorie counts, it's an awkward concept to use in marketing materials.

Instead, let's calculate the percentage of cereals that contain more than 60 calories per serving. Save your answer in the more_calories variable and print it.

```python
more_calories = np.mean(calorie_stats > 60) * 100
print(f"The {more_calories:.0f}% companies cereals have more than 60 calories per serving")
```

7. This is a really high percentage. This will be very useful as we promote CrunchieMunchies. But one question is: how big are the differences in the dataset? Can we generalize that most cereals contain around 100 calories or is the range even wider?

Calculate the magnitude of the deviation by finding the standard deviation. Save your answer in calorie_std and print it on the terminal. How can we incorporate this value into our analysis?

```python
calorie_std = np.std(calorie_stats)

print(f"Standard deviation totals: {calorie_std}")
print(f"Most of companies contain from {round(average_calories-calorie_std)} to {round(average_calories+calorie_std)} calories")
```
